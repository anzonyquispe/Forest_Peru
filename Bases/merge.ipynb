```python
# =========================================================
# 1. IMPORTAR LIBRERÍAS
# =========================================================

import pandas as pd
import os
```

```python
# =========================================================
# 2. DEFINIR CARPETA DE TRABAJO
# =========================================================
# Cambia esta ruta por la carpeta donde tienes TODOS los CSV

ruta = r"/Users/karlavega/Proyectos/Colaborativos/Forest_Peru/Bases"

# Ir a la carpeta
os.chdir(ruta)
```

```python

# =====================================================
# AÑOS
# =====================================================

anios = [2006, 2010, 2014, 2018, 2022]

# =====================================================
# LOOP
# =====================================================

for anio in anios:

    print(f"\n==============================")
    print(f"Procesando año {anio}")
    print(f"==============================")

    # =================================================
    # LEER BASE DISTRITAL
    # =================================================

    distrital = pd.read_csv(
        f"distrital_{anio}.csv",
        encoding="latin1"
    )

    # =================================================
    # LEER BASE ALCALDES
    # =================================================

    alcaldes = pd.read_csv(
        f"alcaldes_con_votos_{anio}.csv",
        encoding="utf-8-sig"
    )

    # =================================================
    # LIMPIAR NOMBRES COLUMNAS
    # =================================================

    distrital.columns = (
        distrital.columns
        .str.strip()
        .str.lower()
        .str.replace("ï»¿", "", regex=False)
    )

    alcaldes.columns = (
        alcaldes.columns
        .str.strip()
        .str.lower()
        .str.replace("ï»¿", "", regex=False)
    )

    # =================================================
    # MOSTRAR COLUMNAS
    # =================================================

    print("\nColumnas distrital:")
    print(distrital.columns.tolist())

    print("\nColumnas alcaldes:")
    print(alcaldes.columns.tolist())

    # =================================================
    # ASEGURAR UBIGEO
    # =================================================

    if "ubigeo" not in distrital.columns:
        print(f"ERROR: distrital_{anio} no tiene ubigeo")
        continue

    if "ubigeo" not in alcaldes.columns:
        print(f"ERROR: alcaldes_{anio} no tiene ubigeo")
        continue

    # =================================================
    # UBIGEO COMO STRING
    # =================================================

    distrital["ubigeo"] = distrital["ubigeo"].astype(str)
    alcaldes["ubigeo"] = alcaldes["ubigeo"].astype(str)

    # =================================================
    # VARIABLES A MERGEAR
    # =================================================

    alcaldes_small = alcaldes[
        ["ubigeo", "sexo", "origen_apellido"]
    ]

    # =================================================
    # ELIMINAR DUPLICADOS
    # =================================================

    alcaldes_small = alcaldes_small.drop_duplicates(
        subset="ubigeo"
    )

    # =================================================
    # MERGE
    # =================================================

    merge_final = distrital.merge(
        alcaldes_small,
        on="ubigeo",
        how="left"
    )

    # =================================================
    # GUARDAR CSV
    # =================================================

    nombre_salida = f"distrital_merge_{anio}.csv"

    merge_final.to_csv(
        nombre_salida,
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\nArchivo guardado:")
    print(nombre_salida)

print("\nProceso terminado correctamente.")
```
