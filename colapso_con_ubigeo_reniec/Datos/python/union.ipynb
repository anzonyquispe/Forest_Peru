```python
import pandas as pd
import numpy as np

# ============================
# 1. Cargar base de resultados distritales
# ============================
# Asegúrate de que el archivo esté en la misma carpeta que este script
res = pd.read_excel("/Users/karlavega/Proyectos/Colaborativos/Forest_Peru/Datos/BD/resultados_distrital.xlsx")

# Nos quedamos solo con 2018 y 2022 (para poder comparar y crear turnover)
res_filtrado = res[res["año"].isin([2018, 2022])].copy()

```

```python
# ============================
# 2. Ganador por ubigeo y año (máximo total_votos)
# ============================
winners = (
    res_filtrado
    .sort_values(["ubigeo", "año", "total_votos"], ascending=[True, True, False])
    .groupby(["ubigeo", "año"], as_index=False)
    .first()
)

# winners tiene columnas:
# ubigeo, año, region, provincia, distrito,
# organizacion_politica, total_votos, orden_aparicion

```
