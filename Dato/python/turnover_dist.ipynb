```python
import pandas as pd

# 1. Cargar la base con ganadores
df = pd.read_excel("/Users/karlavega/Proyectos/Colaborativos/Forest_Peru/Datos/BD/resultados_distrital_con_ganador.xlsx")

# 2. Nos quedamos solo con los ganadores (ganador = 1)
df_ganadores = df[df["ganador"] == 1].copy()

```

```python
# 3. Ordenamos por distrito y año
df_ganadores = df_ganadores.sort_values(["ubigeo", "año"])

# 4. Crear variables de turnover comparando con el ganador del periodo anterior
df_ganadores["org_anterior"] = df_ganadores.groupby("ubigeo")["organizacion_politica"].shift(1)
df_ganadores["candidato_anterior"] = df_ganadores.groupby("ubigeo")["candidato_nombre_completo"].shift(1)

# 5. Variable turnover por organización política
df_ganadores["turnover_org"] = (df_ganadores["organizacion_politica"] != df_ganadores["org_anterior"]).astype(int)

```

```python
# 6. Variable turnover por candidato
df_ganadores["turnover_candidato"] = (df_ganadores["candidato_nombre_completo"] != df_ganadores["candidato_anterior"]).astype(int)

# 7. Filtrar solo resultados del 2022 (el año final requerido)
df_2022 = df_ganadores[df_ganadores["año"] == 2022].copy()
```

```python
# 8. Guardar la base final
df_2022.to_excel("/Users/karlavega/Proyectos/Colaborativos/Forest_Peru/Datos/BD/resultados_turnover_2022.xlsx", index=False)

print("Archivo generado: resultados_turnover_2022.xlsx")
```
