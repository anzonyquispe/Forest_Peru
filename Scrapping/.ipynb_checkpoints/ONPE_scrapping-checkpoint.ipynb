```python
# Exporto las librerías
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver import ActionChains
from selenium.common.exceptions import NoSuchElementException, TimeoutException
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options

import time, random, os, re, json
import pandas as pd
from pandas import DataFrame
from bs4 import BeautifulSoup
import unidecode
from pathlib import Path
```

```python
# Parámetros
UBIGEOS = [
    "010201"  # agrega más: "140407", ...
]
OUT_DIR = os.path.expanduser("~/Desktop/salida_onpe")
BASE_URL = "https://resultadoshistorico.onpe.gob.pe/ERM2022/Actas/Ubigeo"
HEADLESS = True  # pon False si quieres ver el navegador
```

```python
# Inicializar Chrome
options = Options()
if HEADLESS:
    options.add_argument("--headless=new")
options.add_argument("--start-maximized")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(service=webdriver.chrome.service.Service(ChromeDriverManager().install()), options=options)
wait = WebDriverWait(driver, 20)
action = ActionChains(driver)
Path(OUT_DIR).mkdir(parents=True, exist_ok=True)
print("Driver listo. Carpeta de salida:", OUT_DIR)
```

```python
# Utilidades
def _norm_text(t: str) -> str:
    if t is None:
        return ""
    t = t.replace("\xa0", " ")
    import re as _re
    return _re.sub(r"\s+", " ", t).strip()

def _digits_only(s: str) -> str:
    return "".join(ch for ch in (s or "") if ch.isdigit())

def _maybe_click(xpath_candidates):
    for xp in xpath_candidates:
        try:
            el = wait.until(EC.element_to_be_clickable((By.XPATH, xp)))
            el.click()
            return True
        except Exception:
            continue
    return False

def _find_element_any(selectors_css=None, selectors_xpath=None):
    selectors_css = selectors_css or []
    selectors_xpath = selectors_xpath or []
    for sel in selectors_css:
        try:
            return driver.find_element(By.CSS_SELECTOR, sel)
        except Exception:
            pass
    for xp in selectors_xpath:
        try:
            return driver.find_element(By.XPATH, xp)
        except Exception:
            pass
    return None

def _extract_table_rows_from_html(html: str):
    from bs4 import BeautifulSoup
    import re
    soup = BeautifulSoup(html, "lxml")
    rows = []
    # 1) Tablas
    for table in soup.find_all(["table"]):
        body_rows = table.select("tbody tr") or table.select("tr")
        for tr in body_rows:
            cells = [ _norm_text(td.get_text()) for td in tr.select("th,td") ]
            if len(cells) < 2:
                continue
            numeric_idx = [i for i, c in enumerate(cells) if _digits_only(c) != ""]
            if numeric_idx:
                vi = numeric_idx[-1]
                votos = _digits_only(cells[vi])
                org = cells[vi-1] if vi-1 >= 0 else cells[0]
                porc = next((c for c in cells if "%" in c), "")
                if org and votos:
                    rows.append({
                        "organizacion_politica": org,
                        "votos": votos,
                        "porcentaje": porc
                    })
    if rows:
        return rows
    # 2) Fallback bloques
    for block in soup.find_all(["div","li","span","p"]):
        txt = _norm_text(block.get_text())
        if not txt:
            continue
        for line in txt.split("\n"):
            line = _norm_text(line)
            parts = line.split()
            if not parts:
                continue
            if _digits_only(parts[-1]):
                org = _norm_text(" ".join(parts[:-1]))
                votos = _digits_only(parts[-1])
                porc = ""
                if "%" in line:
                    m = re.search(r"(\d+(?:[\.,]\d+)?)%", line)
                    if m:
                        porc = m.group(0)
                if org and votos:
                    rows.append({
                        "organizacion_politica": org,
                        "votos": votos,
                        "porcentaje": porc
                    })
    return rows
```

```python
def scrape_ubigeo(ubigeo: str) -> pd.DataFrame:
    driver.get(BASE_URL)
    try:
        wait.until(lambda d: d.execute_script("return document.readyState") == "complete")
    except Exception:
        pass
    import time as _t
    _t.sleep(1)

    input_box = _find_element_any(
        selectors_css=[
            "input[placeholder*='Ubigeo' i]",
            "input#ubigeo",
            "input[name*='ubigeo' i]",
        ],
        selectors_xpath=[
            "//input[contains(translate(@placeholder,'UBIGEO','ubigeo'),'ubigeo')]",
            "//input[contains(translate(@name,'UBIGEO','ubigeo'),'ubigeo')]",
            "//label[contains(.,'Ubigeo')]/following::input[1]",
        ],
    )
    if input_box is None:
        raise RuntimeError("No se encontró el campo de búsqueda de Ubigeo. Ajusta selectores.")

    input_box.clear()
    input_box.send_keys(ubigeo)
    _t.sleep(0.3)

    clicked = _maybe_click([
        "//button[contains(.,'Buscar') or contains(.,'Consultar') or contains(.,'Ir')]",
        "//span[contains(.,'Buscar') or contains(.,'Consultar') or contains(.,'Ir')]/ancestor::button[1]",
    ])
    if not clicked:
        from selenium.webdriver.common.keys import Keys
        input_box.send_keys(Keys.ENTER)

    _t.sleep(1.0)
    try:
        wait.until(lambda d: d.execute_script("return document.readyState") == "complete")
    except Exception:
        pass
    _t.sleep(0.5)

    html = driver.page_source
    rows = _extract_table_rows_from_html(html)
    if not rows:
        _maybe_click([
            "//button[contains(.,'ver más') or contains(.,'Ver más') or contains(.,'Detalle') or contains(.,'Resultados')]",
        ])
        _t.sleep(0.6)
        html = driver.page_source
        rows = _extract_table_rows_from_html(html)

    if not rows:
        raise RuntimeError("No se hallaron filas de resultados. Identifica los selectores de la tabla en el DOM y ajusta utilidades.")

    df = pd.DataFrame(rows)
    df.insert(0, "ubigeo", ubigeo)
    return df
```

```python
# Ejecutar scraping por bloques
master_frames = []
for u in UBIGEOS:
    try:
        print(f"==> Procesando ubigeo {u}…")
        df = scrape_ubigeo(u)
        out_file = os.path.join(OUT_DIR, f"onpe_erm2022_{u}.csv")
        df.to_csv(out_file, index=False, encoding="utf-8")
        print(f"Guardado {out_file} ({len(df)} filas)")
        master_frames.append(df)
        import time as _t, random as _r
        _t.sleep(_r.uniform(0.5, 1.2))
    except Exception as e:
        print(f"[ERROR] Ubigeo {u}: {e}")

if master_frames:
    master = pd.concat(master_frames, ignore_index=True)
    master_file = os.path.join(OUT_DIR, "onpe_erm2022_master.csv")
    master.to_csv(master_file, index=False, encoding="utf-8")
    print(f"CSV maestro: {master_file} ({len(master)} filas)")
else:
    print("No se generaron resultados. Ajusta selectores y vuelve a ejecutar.")
```

```python
# Cerrar navegador
try:
    driver.quit()
except Exception:
    pass
print("Listo.")
```
